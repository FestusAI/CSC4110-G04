## Task 1(d): Problem Statement

### Problem Description

Mulika SACCO wants to improve how it assesses loan applications and estimates interest rates. Loan officers need a way to identify applicants who may fail to repay their loans. Our group will develop machine learning models using historical loan records to predict interest rates and loan default risk. The models will help loan officers make informed decisions, but they will not replace human judgment.

### Machine Learning Targets

**Regression Target:** `loan_int_rate`

This model predicts the annual interest rate of a loan as a numerical value.

**Classification Target:** `loan_status`

This model predicts whether a loan is likely to default, where 1 represents default and 0 represents non-default.

### Expected Important Predictors

We expect the following features to be useful:

- `person_income`: Indicates the borrower's income and potential repayment capacity.
- `loan_amnt`: Shows the amount of money borrowed.
- `loan_percent_income`: Measures the loan amount relative to the borrower's income.
- `person_emp_length`: Provides information about employment history.
- `cb_person_default_on_file`: Indicates whether a previous default is recorded.
- `cb_person_cred_hist_length`: Shows the length of the borrower's credit history.

These variables may help identify repayment risk and explain differences in loan interest rates. Their actual importance will be evaluated during modelling.

**Important:** We will exclude `loan_grade` and `loan_status` from the interest-rate model's input features to avoid data leakage.

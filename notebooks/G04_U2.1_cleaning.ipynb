
## Task 2.1(b): Data Cleaning

In this section, we remove duplicate records and impossible values, handle missing data, and cap extreme income values using the IQR method.


```python

# Task 2.1(b): Remove duplicates and impossible values

# Work on a copy so the original dataset stays unchanged
df_work = df_raw.copy(deep=True)

# Count and remove duplicate rows
duplicates_removed = df_work.duplicated().sum()
df_work = df_work.drop_duplicates().copy()

print("Duplicate rows removed:", duplicates_removed)

# Identify impossible ages above 100 years
invalid_age = df_work["person_age"] > 100

# Identify impossible employment lengths above 60 years
invalid_employment = df_work["person_emp_length"] > 60

# Count invalid values before removing them
print("Rows with age above 100:", invalid_age.sum())
print("Rows with employment above 60:", invalid_employment.sum())

# Remove rows containing either impossible value
invalid_rows = invalid_age | invalid_employment
invalid_removed = invalid_rows.sum()

df_work = df_work.loc[~invalid_rows].copy()

# Show how many records remain
print("Invalid rows removed:", invalid_removed)
print("Rows remaining after cleaning:", len(df_work))
```

```python

# Task 2.1(b): Handle missing values

# Count missing employment lengths before filling
missing_emp = df_work["person_emp_length"].isna().sum()

# Calculate the median employment length
emp_median = df_work["person_emp_length"].median()

# Replace missing employment lengths with the median
df_work["person_emp_length"] = df_work["person_emp_length"].fillna(emp_median)

print("Missing employment lengths filled:", missing_emp)
print("Median employment length:", emp_median)

# Identify rows whose interest rate was originally missing
df_work["rate_imputed"] = df_work["loan_int_rate"].isna().astype(int)

# Count missing interest rates
missing_rates = df_work["rate_imputed"].sum()

# Find the median interest rate for each loan grade
grade_medians = df_work.groupby("loan_grade")["loan_int_rate"].transform("median")

# Fill missing interest rates using their grade median
df_work["loan_int_rate"] = df_work["loan_int_rate"].fillna(grade_medians)

print("Interest rates filled:", missing_rates)

# Verify that the missing values have been handled
print("Remaining missing employment lengths:",
      df_work["person_emp_length"].isna().sum())

print("Remaining missing interest rates:",
      df_work["loan_int_rate"].isna().sum())

# Check how many rows were flagged
print("Rows marked rate_imputed = 1:",
      df_work["rate_imputed"].sum())
```

```python

# Task 2.1(b): Handle income outliers using IQR

# Select the income column
outlier_col = "person_income"

# Calculate the first and third quartiles
Q1 = df_work[outlier_col].quantile(0.25)
Q3 = df_work[outlier_col].quantile(0.75)

# Calculate the interquartile range
IQR = Q3 - Q1

# Calculate the lower and upper fences
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Count incomes above the upper fence
income_outliers = (df_work[outlier_col] > upper).sum()

# Record the highest income before capping
max_before = df_work[outlier_col].max()

# Cap high incomes at the upper fence
df_work[outlier_col] = df_work[outlier_col].clip(upper=upper)

# Display the results
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower fence:", lower)
print("Upper fence:", upper)
print("Income values capped:", income_outliers)
print("Maximum income before:", max_before)
print("Maximum income after:", df_work[outlier_col].max())
print("Rows remaining:", len(df_work))
```

```python

# Task 2.1(b): Check data types and save cleaned data

# Identify columns that should contain numbers
numeric_columns = [
    "person_age",
    "person_income",
    "person_emp_length",
    "loan_amnt",
    "loan_int_rate",
    "loan_status",
    "loan_percent_income",
    "cb_person_cred_hist_length",
    "rate_imputed"
]

# Ensure numerical columns have numerical data types
for column in numeric_columns:
    df_work[column] = pd.to_numeric(
        df_work[column], errors="raise"
    )

# Store the final cleaned dataset
df_stage1 = df_work.copy(deep=True)

# Save the cleaned dataset to Google Drive
df_stage1.to_csv(
    "data/stage1_credit.csv",
    index=False
)

# Display the results
print("CLEANED DATASET SAVED SUCCESSFULLY")

print("\nFinal dataset shape:")
print(df_stage1.shape)

print("\nFinal data types:")
print(df_stage1.dtypes)

print("\nRemaining missing values:")
print(df_stage1.isnull().sum())

print("\nImputed interest rate flags:")
print(df_stage1["rate_imputed"].value_counts())

print("\nSaved file: data/stage1_credit.csv")
```

The original Credit Risk Dataset contained 32,581 rows and 12 columns. During data cleaning, I made the following decisions:

**1. Removing duplicates**

I removed 165 duplicate records to prevent the same loan from being counted more than once.

**2. Removing impossible values**

I removed 5 rows containing borrower ages above 100 years and 2 rows containing employment lengths above 60 years. These values were unrealistic and could affect the accuracy of our analysis.

**3. Filling missing employment lengths**

I replaced 887 missing employment-length values with the median employment length of 4 years. I used the median because it is less affected by extreme values than the mean.

**4. Filling missing interest rates**

I replaced 3,094 missing interest-rate values with the median interest rate of the corresponding loan grade. I also created a new column named `rate_imputed` to identify these records. The imputed interest rates must be excluded when training the regression models.

**5. Handling income outliers**

I applied the Interquartile Range (IQR) method to the `person_income` column.

- First quartile (Q1): 38,500
- Third quartile (Q3): 79,200
- IQR: 40,700
- Lower fence: -22,550
- Upper fence: 140,250

I capped 1,474 income values above 140,250 at the upper fence. I used capping instead of deleting these records to preserve useful borrower information.

**6. Checking data types**

I checked that numerical columns had appropriate numerical data types. This ensures that statistical calculations and machine learning operations can be performed correctly.

**7. Columns added or removed**

I did not drop any original columns. I added one new column named `rate_imputed` to indicate which interest rates were filled.

**Final result**

After cleaning, the dataset contained 32,409 rows and 13 columns, with no missing values remaining. I stored the cleaned data in the variable `df_stage1` and saved it as `data/stage1_credit.csv` for the next stage of the project.
